```python
#Ejemplo de dos beta1 = 0 en los que existe una correlación (aunque no es lineal)

import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm

# Semilla para reproducibilidad
np.random.seed(123)

# Tiempo
tiempo = np.arange(0, 12.1, 0.1)

# y1: ruido puro
y1 = np.random.normal(0, 3, len(tiempo))

# y2: relación cuadrática + ruido
y2 = (tiempo - 6)**2 + np.random.normal(0, 3, len(tiempo))

# 📊 Gráficos lado a lado
fig, axs = plt.subplots(1, 2, figsize=(12, 5))

axs[0].scatter(tiempo, y1, color='blue', s=30)
axs[0].set_xlabel("x")
axs[0].set_ylabel("y1")
axs[0].set_title("Relación ruido puro")

axs[1].scatter(tiempo, y2, color='green', s=30)
axs[1].set_xlabel("x")
axs[1].set_ylabel("y2")
axs[1].set_title("Relación cuadrática + ruido")

plt.tight_layout()
plt.show()


```

```python
# 📈 Regresión lineal: tiempo ~ y1
X1 = sm.add_constant(y1)
modelo1 = sm.OLS(tiempo, X1).fit()
print("Modelo 1: tiempo ~ y1")
print(modelo1.summary())

```

```python
# 📈 Regresión lineal: tiempo ~ y2
X2 = sm.add_constant(y2)
modelo2 = sm.OLS(tiempo, X2).fit()
print("\nModelo 2: tiempo ~ y2")
print(modelo2.summary())
```

La nota al final es una advertencia técnica que te recuerda lo siguiente:

Los errores estándar que ves en el resumen del modelo (y que se usan para calcular los valores t, p, etc.) asumen que los residuos del modelo cumplen ciertas condiciones:

- Homocedasticidad (varianza constante)

- No autocorrelación

- Independencia

- Normalidad (en algunos contextos)

Si esas condiciones no se cumplen, los errores estándar podrían estar subestimados o sobreestimados, lo que afectaría la significancia estadística de los coeficientes.
