# Estadística Descriptiva: medidas de posición y de dispersión a partir de datasets

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import datasets
```

```python
datos = pd.read_csv('https://raw.githubusercontent.com/mcappelletti/cursos/refs/heads/main/share-of-population-suffering-from-cancer-types.csv')
datos.head()
```

```python
# Filtrar datos de Argentina y España
datos_arg = datos[datos['Entity'] == 'Argentina']
datos_esp = datos[datos['Entity'] == 'Spain']
datos_arg.head()
```

```python
# Visualización con matplotlib/seaborn
fig, axes = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

sns.lineplot(data=datos_arg, x='Year', y='Liver', ax=axes[0])
axes[0].set_title('Argentina - Liver Cancer')
axes[0].set_ylabel('Liver')
axes[0].set_xlabel('Años')

sns.lineplot(data=datos_esp, x='Year', y='Liver', ax=axes[1])
axes[1].set_title('España - Liver Cancer')
axes[1].set_ylabel('Liver')
axes[1].set_xlabel('Años')

plt.tight_layout()
plt.show()
```

```python
# Dataset incorporado: iris
iris = datasets.load_iris()
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
iris_df['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)
iris_df.head()
```
