# Regresión Lineal
---

* Es la técnica estadística para el modelado y la investigación de la relación entre dos o más variables
* Consiste en obtener una función de aproximación que se ajuste a la forma o a la tendencia general de los datos, sin coincidir necesariamente en todos los puntos

## Ajuste de curvas

* Ajustar una curva a los datos experimentales consiste en determinar una fórmula y=f(x) que relacione las variables x e y
* Métodos:
    * Ajuste por regresión lineal en mínimos cuadrados
    * Ajuste por polinomio de grado m
    * Ajuste potencial
    * Ajuste exponencial
    * Ajuste cociente o crecimiento saturado

## Regresión Lineal Simple

### Método de mínimos cuadrados

* Interés en estimar la mejor recta que explica los puntos del diagrama de dispersión
* Un enfoque puede ser minimizar las distancias a la recta:

$$e_i = y_i - (\beta_0 + \beta_1x_i)$$

* Estimaciones:

$$\hat{\beta_1} = \frac{S_{xy}}{S_{xx}}$$
$$\hat{\beta_0} = \bar{y} - \hat{\beta_1}\bar{x}$$


```python
#Regresión lineal simple
n = len(datosPos['tiempo'])
xi = datosPos['tiempo']
yi = datosPos['posicion']
Sxx = np.sum(xi**2) - (np.sum(xi)**2) / n
Sxy = np.sum(xi * yi) - (np.sum(xi) * np.sum(yi)) / n
beta1_est = Sxy / Sxx
beta0_est = np.mean(yi) - beta1_est * np.mean(xi)

print(f'Sxx: {Sxx}')
print(f'Sxy: {Sxy}')

print(f'Estimación de beta0 (Ordenada al origen): {beta0_est}')
print(f'Estimación de beta1 (Pendiente): {beta1_est}')
```

```python
# Usando sklearn
model = LinearRegression().fit(xi.values.reshape(-1, 1), yi)
print(f'Ordenada al origen: {model.intercept_}, Pendiente: {model.coef_[0]}')
```

$$y = 5.58 + 9.70x$$

```python
# Calculo de las desviaciones de beta0 y beta1
SST = np.sum(yi**2) - n * (np.mean(yi)**2)
SSE = SST - beta1_est * Sxy
varSigma_est = SSE / (n - 2)

desvBeta1 = np.sqrt(varSigma_est / Sxx)
desvBeta0 = np.sqrt(varSigma_est * (1 / n + (np.mean(xi)**2) / Sxx))

print(f'Desviación típica de beta0: {desvBeta0}')
print(f'Desviación típica de beta1: {desvBeta1}')
```

$$\sigma(\hat{\beta_0}) = 0.418$$
$$\sigma(\hat{\beta_1}) = 0.357$$

### OLS (Ordinary Least Squares)

* Busca minimizar la suma de los cuadrados de los residuos (la diferencia entre los valores reales y los predichos)

```python
import statsmodels.api as sm

X = datosPos[['tiempo']]
X = sm.add_constant(X)
y = datosPos['posicion']
modelo = sm.OLS(y, X).fit()
modelo.summary()
```

### Test de Hipótesis sobre $\beta_1$

$$H_0: \beta_1 = \beta_{1,0}$$
$$H_1: \beta_1 \neq \beta_{1,0}$$

$$\text{Estadístico de prueba:}$$
$$T = \frac{\hat{\beta_1} - \beta_{1,0}}{\sqrt{\hat{\sigma^2} + S_{xx}}}$$

$$H_0: \beta_1 = 0$$
$$H_1: \beta_1 \neq 0$$

```python
# Valor de estadístico de prueba
T0 = (beta1_est - 0) / np.sqrt(varSigma_est / Sxx)
print(f'T0: {T0}')
p_valor = 2 * (1 - stats.t.cdf(T0, n - 2))
print(f'p-valor de beta1: {p_valor}')
```

* p-valor es muy pequeño, por lo tanto, se puede concluir que la se rechaza
* Esto puede significar que existe una relación lineal entre las variables o que con un polinomio de mayor orden se obtenga un mejor resultado

$$H_0: \beta_1 = 10$$
$$H_1: \beta_1 \neq 10$$

```python
# Valor de estadístico de prueba
T0 = (beta1_est - 10) / np.sqrt(varSigma_est / Sxx)
print(f'T0: {T0}')
p_valor = 2 * (stats.t.cdf(T0, n - 2))
print(f'p-valor de beta1: {p_valor}')
```

* En este caso, el p-valor es: 0,418. Por lo tanto, no se puede rechazar
que sea 10

### Test de Hipótesis sobre $\beta_0$

$$H_0: \beta_1 = \beta_{0,0}$$
$$H_1: \beta_1 \neq \beta_{0,0}$$

$$\text{Estadístico de prueba:}$$
$$T = \frac{\hat{\beta_0} - \beta_{0,0}}{\sqrt{\hat{\sigma^2} [\frac{1}{n} + \frac{\bar{x}^2}{S_{xx}}}]}$$

## Análisis de residuos y coeficiente de determinación

* ¿Cuánto de los datos son explicados por un modelo lineal?

### Análisis de residuos

* Es una herramienta para validar un modelo
* Los residuos son las diferencias entre los valores observados y los valores predichos del modelo

$$e_i = y_i - \hat{y_i} = y_i - (\hat{\beta_0} + \hat{\beta_1}x)$$

* Al analizar los residuos, se busca verificar si se cumplen ciertas **suposiciones clave** que justifican la validez del modelo:

| Supuestos                    |                                                                                        | Verificación                                                             |
| --                           | --                                                                                     | --                                                                       |
| Linealidad                   | la relación entre las variables independientes y la variable dependiente es **lineal** | plt.scatter, sns.residplot                                               |
| Independencia de los errores | los residuos deben ser **independientes entre sí**                                     | test de Durbin-Watson (valores cercanos a 2), gráfico residuos vs. tiempo| 
| Homocedasticidad             | varianza de los residuos debe ser **constante**                                        | gráfico residuos vs. valores ajustados (NO embudo)                       |
| Normalidad de los errores    | los residuos están **normalmente distribuidos**   $e_i \sim N(0, \sigma^2)$            | gráfico residuos vs. valores ajustados (residuos centrados alrededor de la línea horizontal en 0). Histograma. QQ-plot |

```python
import pandas as pd
import numpy as np
import seaborn as sns
import statsmodels.api as sm
import scipy.stats as stats
from scipy.stats import t
import matplotlib.pyplot as plt
from statsmodels.stats.outliers_influence import OLSInfluence
```

```python
trees = pd.read_csv('doc/ia/data/trees.csv')
trees.head()
```

#### Normalidad de los errores (residuos normalmente distribuidos)

```python
# Estimar el modelo lineal: Volume ~ Girth
X = sm.add_constant(trees['Girth'])  # Agregar intercepto
y = trees['Volume']
modelo = sm.OLS(y, X).fit()

# Obtener residuos y residuos estandarizados
influence = OLSInfluence(modelo)
residuos_estandarizados = influence.resid_studentized_internal
valores_ajustados = modelo.fittedvalues

# Graficar residuos estandarizados vs valores ajustados
plt.scatter(valores_ajustados, residuos_estandarizados)
plt.axhline(y=2, linestyle='dashed', linewidth=2, color='black')
plt.axhline(y=-2, linestyle='dashed', linewidth=2, color='black')
plt.axhline(y=0, color='black')
plt.xlabel('Valores ajustados')
plt.ylabel('Residuos estandarizados')
plt.title('Diagnóstico de residuos')
plt.show()
```

* Intervalo de confianza al 95%. En general, se toma el intervalo [-2, 2]
* Si un residuo estandarizado cae dentro de [-2,2], es consistente con el comportamiento esperado de un error, bajo la suposición de normalidad
* Si un residuo estandarizado cae fuera de ese rango, entonces estaría fuera del intervalo de confianza al 95%, y eso sugiere (aunque no prueba) que:
    * Esa observación puede ser atípica (un outlier o valor atípico es una observación que se desvía significativamente del resto de los datos)
    * El modelo puede no ajustarse bien ahí
    * La suposición de normalidad no se cumple en ese punto

#### Homocedasticidad (varianza constante)

```python
# Genero datos aleatoriamente con distribución de Poisson
np.random.seed(123)
x = np.arange(1, 50.5, 0.5)
y = [np.random.poisson(lam=xi) for xi in x]

# Modelo lineal
X = sm.add_constant(x)  # Agrega término independiente
modelo = sm.OLS(y, X).fit()

# DataFrame para graficar
misdatos = pd.DataFrame({'x': x, 'y': y})

# Gráfico
sns.set(style='whitegrid')
plt.figure(figsize=(8, 6))
sns.scatterplot(x='x', y='y', data=misdatos, color='blue')

# Línea de regresión con intervalo de confianza de 95%
preds = modelo.get_prediction(X)
pred_summary = preds.summary_frame(alpha=0.05)

plt.plot(x, modelo.predict(X), color='red', label='Regresión lineal')
plt.fill_between(x, pred_summary['mean_ci_lower'], pred_summary['mean_ci_upper'],
                 color='red', alpha=0.3, label='IC 95%')

plt.xlabel('x', fontsize=12, color='black')
plt.ylabel('y', fontsize=12, color='black')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Gráfico de residuos vs valores ajustados (modelo original)
plt.figure(figsize=(6, 4))
plt.scatter(modelo.fittedvalues, sm.OLS(y, X).fit().get_influence().resid_studentized_internal)
plt.xlabel('Valores ajustados')
plt.ylabel('Residuos estandarizados')
plt.title('Modelo original')
plt.tight_layout()
plt.show()
```

```python
# Transformación de varianza -> para estabilizar la varianza
yz = np.sqrt(y)
xz = x

# Modelo lineal con datos transformados
X2 = sm.add_constant(xz)
modelo2 = sm.OLS(yz, X2).fit()

# Gráfico de residuos vs valores ajustados (modelo transformado)
plt.figure(figsize=(6, 4))
plt.scatter(modelo2.fittedvalues, modelo2.get_influence().resid_studentized_internal)
plt.xlabel('Valores ajustados')
plt.ylabel('Residuos estandarizados')
plt.title('Modelo con sqrt(y)')
plt.tight_layout()
plt.show()
```

### Coeficiente de correlación de Pearson

* Considerar a las variables x e y como aleatorias
* X e Y tendrán una distribución conjunta (bidimensional)
* Se puede definir el coeficiente de correlación de Pearson:

$$\rho = \frac{\sigma_{XY}}{\sigma_X \sigma_Y}$$

* donde $\sigma_{XY}$ es la covarianza
* $\rho$ puede tomar valores entre -1 y 1

### Coeficiente de determinación

* Medida de cuánto de la variabilidad de los datos es explicada por el modelo de reresión lineal

$$ R^2 = 1 - \frac{SS_E}{SS_T}$$

* Por ejemplo, si es 0.85, entonces diremos que el 85% de la variabilidad es explicada por el modelo lineal

```python
X = sm.add_constant(trees["Girth"])
y = trees["Volume"]
modelo = sm.OLS(y, X).fit()
modelo.summary()
```

```python
modelo.rsquared
```

El modelo lineal explica el 93,5% de la variabilidad de los datos
