# Estadística Descriptiva: Histogramas

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
datos = pd.read_csv('https://raw.githubusercontent.com/mcappelletti/cursos/refs/heads/main/Blackmore.csv')
datos.head()
```

```python
datos.describe()
```

```python
# Filtrar pacientes menores de 9 años
datos_pre = datos[(datos['group'] == 'patient') & (datos['age'] < 9)]
datos_patient_menores = datos_pre['exercise']
print(f'Cantidad de pacientes menores de 9 años: {len(datos_patient_menores)}')
```

```python
datos_pre.head()
```

```python
datos_patient_menores.head()
```

```python
# Construcción del histograma
plt.hist(datos_patient_menores, bins=10, color='blue', edgecolor='black')
plt.title('Pacientes < 9 años')
plt.xlabel('Ejercicio Hs/semana')
plt.ylabel('Frecuencia')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Resumen estadístico
print('Resumen estadístico:')
datos_patient_menores.describe()
```
