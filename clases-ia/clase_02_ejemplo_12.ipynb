```python
# Ejemplo de t-test

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm, t
```

```python
# Estilo similar a ggplot2
sns.set(style="whitegrid")

# Genero el vector x
x = np.arange(-4, 4, 0.01)
# fdp de la t de Student con 6 grados de libertad
y_fdp = t.pdf(x, df=6)
# Fda de la t de Student con 6 grados de libertad
y_fda = t.cdf(x, df=6)
```

```python
# Creo el DataFrame para graficar
df = pd.DataFrame({'x': x})

# Grafico las funciones
plt.figure(figsize=(8, 5))
plt.plot(x, norm.pdf(x), label='Normal', color='blue')
plt.plot(x, t.pdf(x, df=6), label='t-student', color='red')
plt.title('Comparación entre Normal y t-Student (df=6)')
plt.xlabel('x')
plt.ylabel('Densidad')
plt.legend()
plt.tight_layout()
plt.show()
```
