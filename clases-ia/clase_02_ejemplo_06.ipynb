# Estadística Descriptiva: Histogramas para distribuciones conocidas: Normal y Chi-cuadrado

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Setea la semilla
np.random.seed(641)
n = 1000

# Genera 1000 números aleatorios N(5,1)
muestra_normal = np.random.normal(loc=5, scale=1, size=n)

# Genera 1000 números con distribución Chi-cuadrada con 4 grados de libertad
muestra_chi = np.random.chisquare(df=4, size=n)

# Crea el DataFrame
datosmios = pd.DataFrame({
    'muestra.normal': muestra_normal,
    'muestra.chi': muestra_chi
})
```

```python
# Histograma + densidad para muestra.normal
plt.figure(figsize=(8, 5))
sns.histplot(datosmios['muestra.normal'],
             binwidth=0.5,
             kde=False,
             stat='density',
             color='blue',
             edgecolor='black',
             alpha=0.2)
sns.kdeplot(datosmios['muestra.normal'], color='red')
plt.xlabel('Muestra.normal')
plt.ylabel('Densidad')
plt.title('Distribución Normal')
plt.tight_layout()
plt.show()
```

```python
# Histograma + densidad para muestra.chi
plt.figure(figsize=(8, 5))
sns.histplot(datosmios['muestra.chi'],
             binwidth=0.5,
             kde=False,
             stat='density',
             color='blue',
             edgecolor='black',
             alpha=0.2)
sns.kdeplot(datosmios['muestra.chi'], color='red')
plt.xlabel('Muestra.chi')
plt.ylabel('Densidad')
plt.title('Distribución Chi-cuadrada')
plt.tight_layout()
plt.show()
```
