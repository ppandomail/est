# Estadística Descriptiva: medidas de posición y de dispersión

```python
import numpy as np
import pandas as pd
from scipy.stats import chi2
from collections import Counter
```

```python
# Parámetros
n = 10000
k = 4      # Grados de libertad

# Genero n números con distribución chi-cuadrado con k grados de libertad
y = chi2.rvs(df=k, size=n)
```

```python
# Media
print(f'Valor poblacional de media: {k}') # Valor teórico
print(f'Valor muestral de media: {np.mean(y)}') # Valor muestral
```

```python
# Mediana
print(f'Valor poblacional de mediana: {k - 2/3}') # Aproximación teórica
print(f'Valor muestral de mediana: {np.median(y)}')
```

```python
# Varianza
print(f'Valor poblacional de la varianza: {2 * k}')
print(f'Valor muestral de la varianza: {np.var(y, ddof=1)}') # ddof=1 para muestra
```

```python
# Desviación estándar
print(f'Valor poblacional de la desviación estándar: {np.sqrt(2 * k)}')
print(f'Valor muestral de la desviación estándar: {np.std(y, ddof=1)}')
```

```python
print(f'Cuantiles: {np.quantile(y, [0, 0.25, 0.5, 0.75, 1])}')
print(f'Rango: {(np.min(y), np.max(y))}')
print(f'Rango intercuartílico: {np.percentile(y, 75) - np.percentile(y, 25)}')
```

```python
# Resumen
print('Las principales medidas de posición y dispersión son:')
pd.Series(y).describe()
```

```python
# Moda (para datos categóricos)
def calculomoda(v):
    return Counter(v).most_common(1)[0][0]

datos = ['nose', 'no', 'nose', 'si', 'si', 'si', 'nose', 'nose']
print(f'Moda del vector categórico: {datos}')
calculomoda(datos)
```
