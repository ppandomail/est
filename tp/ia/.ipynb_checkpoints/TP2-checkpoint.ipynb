* Para realizar cierto tipo de experimento se requiere una temperatura media de laboratorio menor a 20°C.
* Si las temperaturas promedios medidas en 10 instantes diferentes (muestra aleatoria) son: 20.2, 19.7, 20.1, 19.7, 18.9, 19.8, 19.9, 20.4, 20.3, 19.6.
* Suponiendo que la muestra fue tomada de una población normal. ¿Se puede concluir que la temperatura media de laboratorio es menor a 20°C?

Se trata de un test de hipótesis para la media de una población normal con varianza desconocida, por lo que se utilizará una t de Student

## 1. Planteamiento de la hipótesis

* Se quiere determinar si la temperatura media es menor que 20°C

$$H0: \mu = 20$$
$$H1: \mu < 20$$

* Es un contraste unilateral a la izquierda.

## 2. Estadísticos muestrales

```python
datos = [20.2, 19.7, 20.1, 19.7, 18.9, 19.8, 19.9, 20.4, 20.3, 19.6]
```

```python
n_tamanio_muestral = len(datos)
n_tamanio_muestral
```

```python
x_media_muestral = sum(datos)/n_tamanio_muestral
x_media_muestral
```

```python
import numpy as np

s_desviación_tipica_muestral = np.std(datos, ddof=1)
s_desviación_tipica_muestral
```

## 3. Estadístico de prueba

$$t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}}$$

donde $\mu_0 = 20$

```python
import math

mu_0 = 20
numerador = x_media_muestral - mu_0
denominador = (s_desviación_tipica_muestral / math.sqrt(n_tamanio_muestral))
t_estadistico_prueba = numerador / denominador
t_estadistico_prueba
```

```python
gl_grados_libertad = n_tamanio_muestral - 1
gl_grados_libertad
```

## 4. Región crítica

### Valor crítico

```python
from scipy.stats import t

alfa_nivel_significacion = 0.05
valor_critico = t.ppf(alfa_nivel_significacion, gl_grados_libertad)
valor_critico
```

La región crítica es t < -1.833

### P-Valor

```python
p_value = t.cdf(t_estadistico_prueba, gl_grados_libertad)
p_value
```

## 5. Tomar decisión

```python
def tomar_decision_por_valor_critico(estadistico, valor_critico):
    if estadistico > valor_critico:
        return 'No se rechaza H0. El estadístico no cae en la región de rechazo'
    return 'Se rechaza H0. El estadístico no cae en la región de rechazo'

tomar_decision_por_valor_critico(t_estadistico_prueba, valor_critico)
```

```python
def tomar_decision_por_p_value(p_value, alfa_nivel_significacion):
    if p_value > alfa_nivel_significacion:
        return 'No se rechaza H0'
    return 'Se rechaza H0'

tomar_decision_por_p_value(p_value, alfa_nivel_significacion)
```

## 6. Conclusión

* Con un nivel de significación del 5%:
$$\text{No hay evidencia suficiente para concluir que}$$
$$\mu < 20ºC$$
* Aunque la media observada es 19.86ºC, la diferencia respecto de 20ºC es demasiado pequeña en relación con la variabilidad de la muestra
* Respuesta final: No se puede afirmar estadísticamente que la temperatura media del laboratorio sea menor que 20°C.
