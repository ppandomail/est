## Test de hipótesis para la media con varianza conocida

El porcentaje deseado de SiO2 en cierto tipo de cemento aluminoso es 5.5. Para probar si el verdadero promedio de porcentaje es 5.5 para una planta de producción en particular, se analizaron 16 muestras obtenidas de manera independiente. Supongamos que el porcentaje de SiO2 en una muestra está normalmente distribuido con $\sigma$ = 3, y con $\bar{X}$ = 5.25.

**1. a) ¿Indica esto de manera concluyente que el verdadero promedio de porcentaje difiere de 5.5?. Utilice 𝛼 = 0.01**

* X: "porcentaje de SiO2 en cierto tipo de cemento aluminoso"
* X ~ $N(\mu, 3^2)$
* tamaño n = 16
* promedio muestral $\bar{X}$ = 5.25
* estadístico de prueba: $\bar{X} \sim N(\mu, \frac{3^2}{16})$
* Normalizar $\bar{X}$ a $Z_0 = \frac{\bar{X} - \mu}{\sigma / \sqrt{n}}$, donde $Z_0 \sim N(0,1)$ entonces $Z_0 = 0.333$
* $H_0: \mu = 5.5$
* $H_1: \mu \neq 5.5$
* $\alpha = 0.01$ entonces $z_{\alpha / 2} = z_{0.005} = 2.575$
* **Regla de decisión:**
    * Rechazar $H_0$ si $\mid Z_0 \mid > 2.575$
    * Aceptar $H_0$ si $\mid Z_0 \mid \leq 2.575$
* Dado que $Z_0 = 0.333 < 2.575$ entonces hay suficiente evidencia para **aceptar $H_0$**

```python
import scipy.stats as stats
import math

def calcular_alfa_2(alfa):
    zalfa_2_neg = stats.norm.ppf(alfa / 2)
    zalfa_2_pos = stats.norm.ppf(1 - alfa / 2)
    return (zalfa_2_neg, zalfa_2_pos)
    
def calcular_Z0(prom_muestral, mu, varianza, n):
    return abs((prom_muestral - mu) / (varianza / math.sqrt(n)))

def decidir(Z0, z_alfa_2):
    if Z0 > z_alfa_2:
        return f'Dado que Z0 = {Z0} > z_alfa_2 = {z_alfa_2} entonces hay suficiente evidencia para rechazar H0'
    return f'Dado que Z0 = {Z0} <= z_alfa_2 = {z_alfa_2} entonces hay suficiente evidencia para aceptar H0'

Z0 = calcular_Z0(5.25, 5.5, 3, 16)
alfa_2 = calcular_alfa_2(0.01)[1]
decidir(Z0, alfa_2)
```

**1. b) Verifique el resultado con el p-valor**

```python
def decidir_por_pvalor(p_valor, alfa):
    if p_valor < alfa:
        return f'Dado que p_valor = {p_valor} < alfa = {alfa} entonces se rechaza H0'
    return f'Dado que p_valor = {p_valor} >= alfa = {alfa} entonces se acepta H0'

def p_valor(Z0):
    return 2 * (1 - stats.norm.cdf(Z0, 0, 1))

decidir_por_pvalor(p_valor(Z0), 0.01)
```

**1. c) Si el verdadero promedio de porcentaje es $\mu = 5.6$. Calcule el error de tipo II ($\beta(5.6)$) y
la potencia del test ($\pi(5.6)$)**

```python
def get_sigma(varianza, n):
    return varianza / math.sqrt(n)
    
def get_valores_criticos(alfa, mu, sigma):
    crit_pos = stats.norm.ppf(1 - alfa / 2, loc=mu, scale=sigma)
    crit_neg = stats.norm.ppf(alfa / 2, loc=mu, scale=sigma)
    return (crit_neg, crit_pos)

def get_beta_value(crit, mu, sigma):
    return stats.norm.cdf(crit[1], loc=mu, scale=sigma) - stats.norm.cdf(crit[0], loc=mu, scale=sigma)

sigma = get_sigma(3, 16)
beta_value = get_beta_value(get_valores_criticos(0.01, 5.5, sigma), 5.6, sigma)
beta_value
```

```python
def potencia_test(beta_value):
    return 1 - beta_value

potencia_test(beta_value)
```

**1. d) ¿Qué valor de n se requiere para satisfacer $\alpha$ = 0.01 y $\beta$(5.6) = 0.01?**

$Z_{\beta_0} = Z_{0.01} = 2.33$

$$n \geq \frac{(Z_{\alpha/2} + Z_{\beta_0})^2 \sigma^2}{(\mu - \mu_0)^2}$$
$$n \geq \frac{(2.575 + 2.33)^2 3^2}{(5.6 - 5.5)^2}$$
$$n \geq 21654$$
