## Dataset

* Trabajé con un dataset correspondiente al **ranking de lenguajes de programación**, que incluye información sobre popularidad, paradigma, año de creación, creador y tipo de tipado de distintos lenguajes (extraído de TIOBE Index Año: 2026).
* La variable analizada fue **Rating_Pct** (porcentaje de popularidad de cada lenguaje).
* Enlace a los datos: [https://www.kaggle.com/datasets/zkskhurram/programming-language-tiobe-index-february-2026](https://www.kaggle.com/datasets/zkskhurram/programming-language-tiobe-index-february-2026)

## Librerias

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Carga de datos

```python
df = pd.read_csv('data/tiobe_index_feb_2026.csv')
df.head()
```

```python
# Histograma
plt.figure(figsize=(8,5))
plt.hist(df['Rating_Pct'], bins=10, edgecolor='black')
plt.title('Distribución de Rating_Pct')
plt.xlabel('Rating (%)')
plt.ylabel('Frecuencia')
plt.savefig('data/histograma_rating.png', dpi=300, bbox_inches='tight')
plt.show()
```

* Un histograma muestra cómo se distribuyen los valores de una variable numérica. Cada barra representa un intervalo de valores y su altura indica cuántas observaciones pertenecen a dicho intervalo.
* Se observa que la mayoría de los lenguajes poseen porcentajes de popularidad bajos, concentrándose principalmente por debajo del 3%. Sin embargo, existen algunos lenguajes con valores considerablemente más altos, como Python, C y C++, generando una distribución asimétrica hacia la derecha.
* Este comportamiento indica que la popularidad no se distribuye uniformemente entre los lenguajes, sino que unos pocos concentran gran parte del interés de los desarrolladores.

```python
# Diagrama de caja
plt.figure(figsize=(8,3))
sns.boxplot(x=df['Rating_Pct'])
plt.title('Boxplot de Rating_Pct')
plt.savefig('data/boxplot_rating.png', dpi=300, bbox_inches='tight')
plt.show()
```

* El diagrama de caja resume la distribución de una variable mediante:
    * La línea central, que representa la mediana.
    * La caja, que contiene el 50% central de los datos.
    * Los bigotes, que muestran la dispersión de los valores.
    * Los puntos alejados, que representan posibles valores atípicos.
* El boxplot muestra una concentración importante de valores bajos y la presencia de valores atípicos. En particular, Python presenta un porcentaje de popularidad muy superior al resto de los lenguajes, apareciendo como un dato extremo.
* Además, la distancia entre los valores más altos y la mayoría de las observaciones confirma la desigualdad en la distribución de la popularidad.

## Conclusiones

* Los dos gráficos permiten analizar la distribución de la variable Rating_Pct desde perspectivas complementarias. El histograma muestra la forma general de la distribución, mientras que el diagrama de caja facilita la detección de valores atípicos y la dispersión de los datos.
* A partir de ambos gráficos puede concluirse que la popularidad de los lenguajes de programación se encuentra fuertemente concentrada en unos pocos casos destacados, especialmente Python, mientras que la mayoría presenta porcentajes significativamente menores.
